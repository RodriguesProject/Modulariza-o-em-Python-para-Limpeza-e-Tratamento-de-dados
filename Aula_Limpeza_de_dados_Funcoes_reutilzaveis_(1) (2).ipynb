# Limpeza e validação de dados cadastrais

Notebook de uso do módulo versionado `limpeza.py`. As funções preservam os campos originais e criam indicadores derivados. Os dados de exemplo são didáticos.

```python
import json
import pandas as pd
from limpeza import tratar_clientes, gerar_relatorio_clientes

CAMINHO = 'clientes_sujos.csv'
clientes = pd.read_csv(CAMINHO, dtype={'telefone':'string','cpf':'string','email':'string'})
print(clientes.shape)
display(clientes.head())
```

## Aplicação do módulo

A normalização telefônica reconhece o prefixo brasileiro `55` em números de 12 ou 13 dígitos. O CPF é verificado matematicamente pelos dígitos verificadores. Isso não consulta bases oficiais.

```python
tratados = tratar_clientes(clientes)
relatorio = gerar_relatorio_clientes(tratados)
display(tratados)
print(json.dumps(relatorio, ensure_ascii=False, indent=2))
```

## Exportação

Os arquivos abaixo são produzidos no ambiente de execução e não são enviados automaticamente ao GitHub.

```python
tratados.to_csv('clientes_tratados_final.csv', index=False)
with open('relatorio_clientes.json', 'w', encoding='utf-8') as arquivo:
    json.dump(relatorio, arquivo, ensure_ascii=False, indent=2)
```
